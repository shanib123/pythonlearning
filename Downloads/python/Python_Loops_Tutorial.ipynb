## 1. Why do we need loops?

Imagine printing "Hello" 5 times. Without loops:
```python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
```
This is repetitive. **Loops** let us repeat a block of code without rewriting it.

Python has two main loops:
- **`for` loop** → used when we know how many times to repeat, or when we want to go through items one by one.
- **`while` loop** → used when we repeat *until a condition becomes False* (number of repetitions may not be known in advance).

````python
# Without loop (repetitive)
print("Hello")
print("Hello")
print("Hello")
````

````python
range(3)
````

````python
# With loop (clean)
for i in range(10):
    print("Hello")
````

## 2. `for` loop basics — using `range()`

`range(start, stop, step)` generates numbers. `stop` is **not included**.

- `range(5)` → 0,1,2,3,4
- `range(2,6)` → 2,3,4,5
- `range(0,10,2)` → 0,2,4,6,8

````python
# basic for loop with range
for i in range(5):
    print("Count:", i)
````

````python
# for loop with start, stop
for i in range(2, 6):
    print(i)
````

````python
# for loop with step (using arithmetic operator too)
for i in range(0, 10):
  print(i, "-> square is", i * i)   # using ** or * operator
````

````python
# for loop + conditional statement -> print even/odd
for i in range(1, 11):
    if i % 2 == 0:          # modulus operator
        print(i, "is Even")
    else:
        print(i, "is Odd")
````

````python
# multiplication table of 5
for i in range(1,11):
  print(i,"x 5 =",i*5)
````

````python
for i in range(10):
  print(i,end=" ")
````

````python
# Practical: count vowels in a string using for loop + conditional + list
word = "Programming"
vowels = "aeiouAEIOU"
count = 0

for ch in word:
    if ch in vowels:
        count += 1          # same as count = count + 1
print("Number of vowels in", word, "=", count)
````

````python
# for loop over a dictionary
marks = {"Math": 90, "Science": 85, "English": 78}

for subject in marks:
    print(subject, "->", marks[subject])
print("----------------------------------")
# or using .items()
for subject, score in marks.items():
    print(subject, "scored", score)
````

## 4. `while` loop basics

`while condition:` keeps running the block **as long as the condition is True**.

⚠️ Make sure the condition eventually becomes False, otherwise it becomes an **infinite loop**.

````python
# basic while loop - countdown
n = 10
while n > 0:
    print(n)
    n -= 1          # important: update the variable, else infinite loop
print("Liftoff!")
````

````python
# while loop with input - keep asking until user enters 'stop'
# (Run this cell and type a few words, then type 'stop')
word = ""
while word.lower() != "stop":
    word = input("Enter a word (type 'stop' to end): ")
````

````python
# Practical: sum of numbers entered by user until they enter 0
total = 0
num = int(input("Enter a number (0 to stop): "))

while num != 0:
    total += num
    num = int(input("Enter a number (0 to stop): "))

print("Total sum =", total)
````

## 5. `break` and `continue`

- **`break`** → stops the loop immediately.
- **`continue`** → skips the current iteration and moves to the next one.

````python
# break example - stop loop when we find the number
numbers = [4, 8, 15, 16, 23, 42]
search = 16

for n in numbers:
    if n == search:
        print("Found", search, "in the list!")
        break
    print("Checked", n, "- not a match")
````

````python
# continue example - skip printing negative numbers
numbers = [10, -5, 20, -3, 8]

for n in numbers:
    if n < 0:
        continue        # skip rest of code for this value
    print(n)
````

## 6. Nested loops (loop inside a loop)

Useful for grids, tables, and patterns.

````python
# Outer loop runs 3 times
for i in range(3):

    # Inner loop runs 4 times for every outer loop iteration
    for j in range(4):
        print(j)
````

````python
# Outer loop controls the number of rows
for i in range(3):

    # Inner loop controls the number of columns
    for j in range(4):
        print("*", end=" ")

    # Move to the next line after completing one row
    print()
````

````python
# multiplication table using nested for loops
for i in range(1,11):
    for j in range(1,11):
        print(i,"X",j,'=',i*j)
    print("-----")
````

````python
# Number pyramid — each row prints DIFFERENT numbers
for i in range(1,6):
    for j in range(1,i+1):
        print(j, end =" ")
    print()  
````

````python
#  Hollow square — border is '*', inside is blank (needs per-position decision)
n = int(input("enter a number"))
for i in range(n):
    for j in range(n):
        if i==0 or i==n-1 or j==0 or j==n-1:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
````

````python
for i in range(7,1,-1):
    for j in range(1,i-1):
        print(j, end=" ")
    print()
````

````python
# Practical 2: Check if a number is prime (loops + conditionals + operators)
num = int(input("enter a number"))
is_prime =True

if num<2:
    is_prime = False
else:
    for i in range(2,num):
        if num%i==0:
            is_prime = False
            break

    if is_prime:
        print(num,'is a prime number')
    else:
        print(num,'not a prime number')
````
