## 1. What is a Tuple?

A **tuple** is an ordered collection of items written using **parentheses** `( )`.

- Items are separated by commas
- Items keep their order (indexed from `0`)
- A tuple **cannot be changed** after creation
- Can hold mixed data types

````python
# Simple tuple
fruits = ("apple", "banana", "mango")
print(fruits)

# Mixed data types
student = ("Aisha", 21, 8.5)
print(student)

# Number tuple
scores = (95, 88, 72, 91)
print(scores)

# Check the type
print(type(fruits))
````

## 2. Accessing Items in a Tuple

Use the **index number** to access items. Indexing starts from `0`.

```
fruits = ( "apple" , "banana" , "mango" )
index      0          1           2
negative  -3         -2          -1
```

**Slicing:** `tuple[start:end]` — gets items from start up to (not including) end.

````python
fruits = ("apple", "banana", "mango", "grape", "kiwi")

# Positive indexing
print(fruits[0])
print(fruits[2])

# Negative indexing
print(fruits[-1])
print(fruits[-2])

# Slicing
print(fruits[1:4])
print(fruits[:3])
print(fruits[2:])
````

## 3. Tuple Methods

Tuples have only **2 built-in methods**:

| Method | What it does |
|--------|--------------|
| `count(x)` | Returns how many times `x` appears |
| `index(x)` | Returns the first position of `x` |

Other useful functions: `len()`, `max()`, `min()`, `sum()`

````python
marks = (85, 90, 75, 90, 88, 90, 75)

# count() — how many times does 90 appear?
print(marks.count(90))

# index() — where does 75 first appear?
print(marks.index(75))

# Built-in functions
print("Length :", len(marks))
print("Maximum:", max(marks))
print("Minimum:", min(marks))
print("Total  :", sum(marks))
````

## 4. Immutable vs Mutable Data Types

| Term | Meaning | Examples |
|------|---------|----------|
| **Immutable** | Cannot be changed after creation | Tuple, String, int, float |
| **Mutable** | Can be changed after creation | List, Dictionary, Set |

Think of a **tuple** like a sealed box — you can look inside but cannot change what's in it.  
A **list** is like an open basket — you can add, remove, or swap items freely.

````python
# LIST is MUTABLE — we can change items
fruits_list = ["apple", "banana", "mango"]
print("Before:", fruits_list)

fruits_list[1] = "grapes"   # Change second item
print("After :", fruits_list)
````

````python
# TUPLE is IMMUTABLE — items CANNOT be changed
fruits_tuple = ("apple", "banana", "mango")
print(fruits_tuple)

# Uncomment the line below to see a TypeError:
# fruits_tuple[1] = "grapes"
````

````python
# STRING is also IMMUTABLE
name = "Aisha"
print(name[0])   # Can read a character — OK

# Uncomment below to see TypeError:
# name[0] = "B"

# Good use cases for tuples (data that should never change)
days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
location = (10.52, 76.21)   # Latitude, Longitude
rgb_red  = (255, 0, 0)      # Colour value

print(days)
print(location)
print(rgb_red)
````

## 5. List Creation

A **list** is ordered, mutable, and written with **square brackets** `[ ]`.

There are several ways to create a list:

````python
# Method 1: Direct creation
names = ["Riya", "Arjun", "Meera"]
print("Method 1:", names)

# Method 2: Start empty and use append()
numbers = []
numbers.append(10)
numbers.append(20)
numbers.append(30)
print("Method 2:", numbers)

# Method 3: Using list() on a string
chars = list("hello")
print("Method 3:", chars)

# Method 4: Using range()
evens = list(range(2, 11, 2))
print("Method 4:", evens)
````

````python
# Converting between list and tuple
my_list  = [1, 2, 3, 4, 5]
my_tuple = (10, 20, 30, 40, 50)

print(type(my_list),  my_list)
print(type(my_tuple), my_tuple)

# List → Tuple
new_tuple = tuple(my_list)
print("List to tuple:", new_tuple)

# Tuple → List
new_list = list(my_tuple)
print("Tuple to list:", new_list)
````

## Summary

| Feature | Tuple `( )` | List `[ ]` |
|---------|------------|----------|
| Syntax | `(1, 2, 3)` | `[1, 2, 3]` |
| Mutable? | No | Yes |
| Methods | `count()`, `index()` | `append()`, `remove()`, etc. |
| Best used when | Data must stay fixed | Data will change |

### Key Takeaways
- **Immutable** types (tuple, string, int) cannot be changed after creation
- **Mutable** types (list, dict, set) can be changed freely
- Use `count()` to find frequency, `index()` to find position in a tuple
- You can convert between list and tuple using `list()` and `tuple()`
