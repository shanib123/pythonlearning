###12. Take a number and **reverse the number**.
    Example: `1234 → 4321`

```python
n = int(input("enter a number"))
reverse = 0

while n>0:
    digits = n%10
    reverse = reverse * 10 + digits
    n = n//10
print("reversed number " , reverse)
```

###18. Take two numbers from the user and find their **GCD/HCF** using a loop.


```python
a = int(input("enter the first number"))
b = int(input("enter the second number"))
gcd = 1
for i in range(1,min(a,b)+1):
    if a%i == 0 and b%i == 0:
        gcd = i

print("GCD/HCF " , gcd)
        
```

###19. Take a number and determine whether it is a **perfect number**.
    Example: `6 → 1 + 2 + 3 = 6`

```python
n = int(input("enter a number"))
total = 0
for i in range(1,n):
    if n%i == 0:
        total = total + i

if total == n:
    print(n, "perfect number")
else:
    print(n, "not a perfect number")
```

###20. Generate the **Fibonacci series** for `n` terms.
    Example: `0 1 1 2 3 5 8 13`

```python
n = int(input("enter a number"))
a = 0
b = 1
for i in range(n):
    print(a, end=" ")
    c = a+b
    a = b
    b = c
```
