###7. A teacher wants to combine two attendance sheets recorded separately as lists:
morning = ["Ana", "Ben", "Cara", "Ana"] and afternoon = ["Ben", "Dev", "Cara"]. Print
the names of students who attended only in the morning.

```python
morning = ["Ana", "Ben", "Cara", "Ana"]
afternoon = ["Ben", "Dev", "Cara"]
morning_set = set(morning)
afternoon_set = set(afternoon)
only_morning = (morning_set - afternoon_set)
print(only_morning)
```

###A quiz has participants from two rounds: round1 = ["Alex", "Bella", "Chris", "Alex"]
