# 📂 File Handling in Python

File handling allows us to **create, read, update, and delete files** using Python.

## 1️⃣ Opening a File

We use the `open()` function:

```python
file = open("example.txt", "r")
```

### 2. **File Modes in Python**

When opening a file, you tell Python *how* you want to use it:

| Mode  | Meaning                                              |
| ----- | ---------------------------------------------------- |
| `"r"` | Read (default mode). File **must exist**.            |
| `"w"` | Write. **Overwrites** the file or creates it.        |
| `"a"` | Append. Adds data to the end. Creates if not exists. |
| `"x"` | Create a new file. **Fails if the file already exists** (raises error). |

---

👉 Always **close the file** after work:

```python
file.close()
```

### 3. **Opening and Reading a File**

```python
file = open("example.txt", "r")  # Open file in read mode
content = file.read()           # Read entire content
print(content)
file.close()                    # Always close after use
```

### 👉 **Using `with` Statement (Best Practice)**

*The `with` statement handles closing automatically:*

```python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)

# File is automatically closed here
```

### 4. **Reading Line by Line**


```python
with open("example.txt", "r") as file:
    lines = file.readlines()  # Returns list of lines
print(lines)
```

```python
with open("example.txt", "r") as file:
    line1 = file.readline() # Read first line only
    print(line1)
```

### 5. **Writing to a File (`"w"` mode)**


```python
with open("output.txt", "w") as file:
    file.write("Lokah Chapter 1: Chandra")

#  Warning: "w" will delete previous content.
```

### 6. **Appending to a File (`"a"` mode)**


```python
with open("output.txt", "a") as file:
    file.write("This line is added at the end.\n")
```

### ✅ Summary:

* Use `"r"` to read, `"w"` to write, `"a"` to append.
* Use `with` to manage files safely.
* Read with `.read()`, `.readlines()`, or line-by-line with `for`.

```python
# Count how many words are in notes.txt.
with open("notes.txt") as file:
    a = file.read()
    b = a.split()
print("count of words = ",len(b))
```

```python
# Count how many characters are in a file.
with open("notes.txt") as f:
    a = f.read()
print("count of words =",len(a))
```

```python
# Combine two files (a.txt and b.txt) into one file c.txt.
with open("a.txt","w+") as f:
    f.write("a b c d e f g h i j k l m n o\n")
    f.seek(0)
    print(f.read)
with open("b.txt","w+") as f:
    f.write("123456789/n")
    f.seek(0)
    print(f.read())
```

```python
with open("a.txt","r") as f:
    a = f.read()
with open("b.txt","r") as f:
    b = f.read()
with open("c.txt","w+") as f:
    f.write(a)
    f.write(b)
    f.seek(0)
    print(f.read())
```
