###6. Build an ATM PIN checker: keep asking for a PIN until it's correct, but lock the user out after 3
wrong attempts.

```python
correct_pin = 1234
attempt = 0

while attempt <3:
    pin = (input("enter your correct pin"))
    if pin == correct_pin:
        print("PIN corrected , access granted")
        break
    else:
        attempt +=1
        print("wrong pin")

if attempt == 3:
    print("invalid PIN, screen is locked")
```

###7. Count how many uppercase letters, lowercase letters, and digits are present in a given string.

```python
text = input("enter a word")
print(text)
uppercase = 0
lowercase = 0
digits = 0

for i in text:
    if i.isupper():
        uppercase +=1
    elif i.islower():
        lowercase +=1
    elif i.isdigit():
        digits +=1

print("count of uppercase letters = ",uppercase)
print("count of lowercase letters = ",lowercase)
print("count of digits = ",digits)

```

###8. Given a sentence, find and print the longest word in it.

```python
sentence = input("enter a longest")

words = sentence.split()
longest = words[0]

for word in words:
    if len(word)>len(longest):
        longest = word

print("longest word = ", longest)
```

###9. Given a list of numbers, split them into two separate lists — one for even numbers, one for odd
numbers.

```python
def seperate_numbers(numbers):
    even = []
    odd = []

    for num in numbers:
        if num%2 == 0:
            even.append(num)
        else:
            odd.append(num)
    return even,odd

numbers = [12,22,33,41,55,62,78]
even,odd = seperate_numbers(numbers)

print("even number : ",even)
print("odd numbers : ",odd)
```

###10. Given a list of numbers, find the second largest number without using sorted() or max().

```python
def second_largest(number):
    largest = 0
    second = 0
    for num in number:
        if num>largest:
            second = largest
            largest = num
        elif num>second and num!=largest:
            second = num

    return second

number=[1,22,23,21,55,66,53]
print(second_largest(number))
    
    
```

###11. Simulate rolling a die 100 times using random numbers, count how many times each face
(1–6) appears, and print the frequency of each.

```python
import random

def roll_dice():
    frequency=[0,0,0,0,0,0]

    for i in range(100):
        number = random.randint(1,6)
        frequency[number - 1] += 1

    for i in range(6):
        print(i+1,"appeared",frequency[i],"times")

roll_dice()
```

###12. The computer picks a random number between 1 and 50. The user keeps guessing, and the

###14. Reverse the digits of a number (e.g., 1234 → 4321) without converting it into a string.

```python
def reversed_digits(n):
    result = 0

    while n>0:
        digits = n%10
        result = result*10+digits
        n = n//10
    return result

n = int(input("enter the number"))
print("reversed number",reversed_digits(n))
```
