# Comprehensions in Python

A comprehension is a short way to build a new list, set or dictionary from existing data, all in a single line. Anything you can do with a small loop that collects results, you can usually write as a comprehension.

## The same task, two ways


```python
squares = []

for n in range(1, 6):
    squares.append(n * n)

print(squares)
```

```python
squares = [n * n for n in range(1, 6)]

print(squares)
```

**How to read it:** `[ what_to_keep   for item in source ]` — "give me this, for every item in that".

## Changing every item

```python
numbers = [12,34,2,8]
doubled = {n * 2 for n in numbers}

print(doubled)
```

```python
fruits = ["apple", "mango", "kiwi"]
loud = [fruit.upper() for fruit in fruits]

print(loud)
```

## Keeping only some items

Add an `if` at the end to keep only the items that pass a test.

```python
ages = [12, 25, 17, 30, 16]
adults = [i for i in ages if i >= 18]

print(adults)
```

```python
even_numbers = [n for n in range(1, 11) if n % 2 == 0]

print(even_numbers)
```

## Choosing between two values

When every item should produce *something*, put the `if ... else ...` at the front instead.

```python
numbers = [1, 2, 3, 4, 5]
labels = ["Even" if n % 2 == 0 else "Odd" for n in numbers]

print(labels)
```

## Working with text

```python
letters = [letter for letter in "banana" if letter != "a"]

print(letters)
```

```python
words = ["sun", "moon", "star"]
lengths = [len(word) for word in words]

print(lengths)
```

## Set comprehension

Same idea with curly braces — the result is a set, so duplicates disappear automatically.

```python
words = ["cat", "horse", "dog", "tiger", "ox"]
unique_lengths = {len(word) for word in words}

print(unique_lengths)
```

## Dictionary comprehension

Curly braces with a `key: value` pair at the front.

```python
numbers = [23, 45,12,45]
cubes = {n: n ** 3 for n in numbers}

print(cubes)
```

```python
marks = {"Ann": 35, "Raj": 72, "Mia": 58}
passed = {name: score for name, score in marks.items() if score >= 40}

print(passed)
```

**A good habit:** if a comprehension starts getting long or hard to read, switch back to a normal loop. Shorter is only better when it's still easy to understand.

## Practice Questions

Start from Q1 and work down — the questions get progressively harder.

**Q1.** A shop's item prices are `[250, 400, 120, 90]`. Create a new list where every price has 10% added to it.

```python
prices = [250, 400, 120, 90]
b = [n*90/100 for n in prices]
print(b)
```

**Q2.** A list of names is `["asha", "ravi", "meera", "john"]`. Create a new list where every name starts with a capital letter.

```python
names = ["asha", "ravi", "meera", "john"]
b = [n.capitalize() for n in names]
print(b)
```

**Q3.** Create a list of all the numbers from 1 to 20 that are divisible by 3.

```python
a = [i for i in range(1,21) if i%3==0 ]
print(a)
```

**Q4.** Exam marks are `[45, 78, 32, 90, 66, 28]`. Create a list of only the marks that are 40 or above.

```python
marks = [45, 78, 32, 90, 66, 28]
a = [ i for i in marks if i>=40]
print(a)
```

**Q5.** Temperatures in Celsius are `[0, 25, 37, 100]`. Create a list of the same temperatures in Fahrenheit (multiply by 9/5, then add 32).

```python
celsius = [0,25,37,100]
b = [i*9/5+32 for i in celsius]
print(b)
```

**Q6.** A list of words is `["data", "science", "is", "fun", "and", "powerful"]`. Create a list of only the words with more than 3 letters, all in capital letters.

```python
words = ["data", "science", "is", "fun", "and","powerful"]
b = [i.upper() for i in words if len(i)>3]
print(b)
```

**Q7.** Test scores are `[55, 82, 39, 91, 47]`. Create a list that says `"Pass"` for each score of 50 or above and `"Fail"` for the rest.

```python
scores = [55, 82, 39, 91, 47]
b = [ "pass" if i>=50 else "fail" for i in scores]
print(b)
```

**Q8.** A sentence is `"the quick brown fox jumps over the lazy dog"`. Create a list with the length of every word in it.

```python
a = "the quick brown fox jumps over the lazy dog"
b = [len(i) for i in a.split()]
print(b)
```

**Q9.** A list of fruits is `["apple", "avocado", "banana", "blueberry", "cherry"]`. Create a set of the distinct first letters.

```python
fruits = ["apple", "avocado", "banana", "blueberry", "cherry"]
b = [i[0] for i in fruits]
print(b)
```

**Q10.** A list of cities is `["Kochi", "Chennai", "Delhi"]`. Create a dictionary where each city is paired with the number of letters in its name.

```python
cities = ["Kochi", "Chennai", "Delhi"]
b = {i:len(i) for i in cities}
print(b)
```

**Q11.** A stock record is `{"pen": 40, "book": 0, "bag": 12, "ruler": 0}`. Create a new dictionary that keeps only the items currently in stock.

```python
stock = {"pen":40, "book":0, "bag": 12, "ruler":0}
b = {i for i,j in stock.items() if j>0}
print(b)
```

**Q12.** A number is stored as the text `"58392"`. Create a list of the squares of only its odd digits.

```python
text = "58392"
b = [int(i)**2 for i in text if int(i)%2!=0 ]
print(b)
```
